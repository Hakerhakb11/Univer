```python
import numpy
import pandas

seq = numpy.linspace(0, 25, 20)
s = pandas.Series(seq)

ratios = s / s.shift(-1)

filtered_ratios = ratios[ratios <= 0.85]

mean_ratio = filtered_ratios.mean()
result_1 = round(mean_ratio, 3)

print(f'Задание 1 Среднее значение: {result_1}\n')
```
